```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

# Load and view your data
shopping_data = pd.read_csv("online_shopping_session_data.csv")
shopping_data = shopping_data.dropna(subset=['CustomerType', 'Purchase'])
shopping_data_filtered = shopping_data[(shopping_data['Month'] == "Nov") | (shopping_data['Month'] == "Dec")]
purchase_rates = shopping_data_filtered.groupby("CustomerType")['Purchase'].mean().to_dict()
shopping_data_filtered_return = shopping_data_filtered[shopping_data_filtered['CustomerType']=="Returning_Customer"]
corr_matrix=shopping_data_filtered_return[['Administrative_Duration','Informational_Duration','ProductRelated_Duration']].corr()
np.fill_diagonal(corr_matrix.values, np.nan)
pairs=corr_matrix.unstack()
pairs=pairs.dropna()
top_pair=pairs.idxmax()
top_value=pairs.max()

top_correlation = {"pair": top_pair, "correlation": top_value}

shopping_data_return=shopping_data[shopping_data['CustomerType']=='Returning_Customer']

boosted_rate=shopping_data_filtered_return['Purchase'].mean()*1.15
from scipy.stats import binom
prob_at_least_100_sales = 1 - binom.cdf(99,n=500, p=boosted_rate)
print(shopping_data_return['Purchase'].mean())
print(boosted_rate)
print(prob_at_least_100_sales)

shopping_data['Purchase'].unique()
```
