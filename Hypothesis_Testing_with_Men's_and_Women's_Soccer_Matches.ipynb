```python
import pandas as pd
womens=pd.read_csv('women_results.csv')
mens=pd.read_csv('men_results.csv')
womens['date']=pd.to_datetime(womens['date'])
mens['date']=pd.to_datetime(mens['date'])
is_worldcup_w = womens['tournament'] == 'FIFA World Cup'
is_date_w = womens['date'] >= "2002-01-01"
womens_filtered = womens[is_worldcup_w & is_date_w]
is_worldcup_m = mens['tournament'] == 'FIFA World Cup'
is_date_m = mens['date'] >= "2002-01-01"
mens_filtered = mens[is_worldcup_m & is_date_m]
womens_filtered['total'] = womens_filtered['home_score'] + womens_filtered['away_score']
mens_filtered['total'] = mens_filtered['home_score'] + mens_filtered['away_score']

import pingouin as pg

t_test=pg.ttest(x=womens_filtered['total'],y=mens_filtered['total'], paired = False)
p_val=t_test['p-val'].iloc[0]
alpha = 0.1
if p_val <= alpha:
    result = "reject"
else:
    result = "fail to reject"

result_dict = {"p_val": p_val, "result": result}
```
