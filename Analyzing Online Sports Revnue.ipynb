```python
# Importing libraries
import pandas as pd
import numpy as np

# Loading the data
brands = pd.read_csv("brands.csv")
finance = pd.read_csv("finance.csv")
info = pd.read_csv("info.csv")
reviews = pd.read_csv("reviews.csv")

# Start coding here...
merged=brands.merge(finance, on='product_id', how='left').merge(info, on='product_id', how='left')
merged.dropna(inplace=True)

merged['price_label']=pd.qcut(
    merged['listing_price'],
    q=4,
    labels=['Budget','Average','Expensive','Elite']
)
merged.head()
adidas_vs_nike=round(merged.groupby(['brand','price_label']).agg(num_products=('product_id','count'),mean_revenue=('revenue','mean')).reset_index(),2)

merger = info.merge(reviews, on='product_id', how='left')
merger.dropna(inplace=True)
merger.head()
merger['desc_len']=merger['description'].str.len()
max_len = merger['desc_len'].max()
upper_bound = int(np.ceil(max_len/100)) * 100
bins=list(range(0,upper_bound+100, 100))
labels=[str(x) for x in range(100,upper_bound+100, 100)]
merger['description_length']=pd.cut(merger['desc_len'],bins=bins, labels=labels).astype(str)
description_lengths=round(merger.groupby('description_length').agg(mean_rating=('rating','mean'),total_reviews=('reviews','sum')).reset_index(),2)
print(description_lengths)
```
