```python
SELECT a.assignment_name, a.region, ROUND(SUM(d.amount),2) AS rounded_total_donation_amount, d1.donor_type
FROM donations AS d
LEFT JOIN assignments AS a ON d.assignment_id = a.assignment_id
LEFT JOIN donors AS d1 ON d.donor_id = d1.donor_id
GROUP BY a.assignment_name, a.region, d1.donor_type
ORDER BY rounded_total_donation_amount DESC
LIMIT 5;

WITH rankings AS (
SELECT assignment_id, impact_score, region, ROW_NUMBER() OVER(PARTITION BY REGION ORDER BY impact_score DESC) AS rank_region
FROM assignments
)
SELECT a.assignment_name, a.region, a.impact_score, COUNT(d.donation_id) AS num_total_donations
FROM assignments AS a
LEFT JOIN donations AS d ON d.assignment_id = a.assignment_id
LEFT JOIN rankings ON rankings.assignment_id = a.assignment_id
WHERE rankings.rank_region = 1
	GROUP BY a.assignment_name, a.region, a.impact_score
ORDER BY a.region ASC;
```
