```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
crimes = pd.read_csv("crimes.csv", dtype={"TIME OCC": str})
crimes['crime_hour']=crimes['TIME OCC'].str[:2].astype(int)
peak_crime_hour = crimes['crime_hour'].value_counts().idxmax()

crimes['d/n']=np.select(
    [(crimes['crime_hour']>21) | (crimes['crime_hour']<4)],
    ['night'],
    default='day'
)

peak_night_crime_location = crimes[crimes['d/n']=='night']['AREA NAME'].value_counts().idxmax()

crimes['age_group']=pd.cut(crimes['Vict Age'], bins=[-1,17,25,34,44,54,64,500], labels=["0-17","18-25","26-34","35-44","45-54","55-64","65+"])

victim_ages = crimes['age_group'].value_counts()
```
